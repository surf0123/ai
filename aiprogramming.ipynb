```python
import tensorflow as tf
from tensorflow.keras import models, layers
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
train_images = x_train / 255.0
test_images = x_test / 255.0

model = models.Sequential()
model.add(layers.Flatten(input_shape=(28, 28)))  # 784차원 변환
model.add(layers.Dense(10, activation='softmax'))  # 출력층(단층)

model.compile(optimizer='adam',loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history=model.fit(train_images, y_train, epochs=5)

test_loss, test_acc = model.evaluate(test_images,  y_test)
print('테스트 정확도:', test_acc)

loss = history.history['loss']
acc = history.history['accuracy']
epochs = range(1, len(loss)+1)

plt.plot(epochs, loss, 'b', label='Training Loss')
plt.plot(epochs, acc, 'r', label='Accuracy')
plt.xlabel('epochs')
plt.ylabel('loss/acc')
plt.show()
#0~9999
pred = model.predict(test_images[0].reshape(1,28,28,1), batch_size=1)
print("추정된 숫자=", pred.argmax())
plt.imshow(test_images[0], cmap="Greys")
```

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

(train_images, train_labels), (test_images, test_labels) = datasets.mnist.load_data()

train_images = train_images.reshape((60000, 28, 28, 1))
test_images = test_images.reshape((10000, 28, 28, 1))

train_images, test_images = train_images / 255.0, test_images / 255.0

model = models.Sequential()
model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.Flatten())
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(10, activation='softmax'))
model.summary()

import matplotlib.pyplot as plt

model.compile(optimizer='adam',loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history=model.fit(train_images, train_labels, epochs=5)

test_loss, test_acc = model.evaluate(test_images, test_labels)
print('테스트 정확도:', test_acc)

loss = history.history['loss']
acc = history.history['accuracy']
epochs = range(1, len(loss)+1)

plt.plot(epochs, loss, 'b', label='Training Loss')
plt.plot(epochs, acc, 'r', label='Accuracy')
plt.xlabel('epochs')
plt.ylabel('loss/acc')
plt.show()

pred = model.predict(test_images[0].reshape(1,28,28,1), batch_size=1)
print("추정된 숫자=", pred.argmax())
plt.imshow(test_images[0], cmap="Greys")

```

LeNet

```python
import numpy as np
from tensorflow import keras
from tensorflow.keras import layers

# Model / data parameters
num_classes = 10
input_shape = (28, 28, 1)

# the data, split between train and test sets
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Scale images to the [0, 1] range
x_train = x_train.astype("float32") / 255
x_test = x_test.astype("float32") / 255
# Make sure images have shape (28, 28, 1)
x_train = np.expand_dims(x_train, -1)
x_test = np.expand_dims(x_test, -1)
print("x_train shape:", x_train.shape)
print(x_train.shape[0], "train samples")
print(x_test.shape[0], "test samples")

# convert class vectors to binary class matrices
y_train = keras.utils.to_categorical(y_train, num_classes)
y_test = keras.utils.to_categorical(y_test, num_classes)

model = keras.Sequential(
    [
        keras.Input(shape=input_shape),
        layers.Conv2D(32, kernel_size=(3, 3), activation="relu"),
        layers.MaxPooling2D(pool_size=(2, 2)),
        layers.Conv2D(64, kernel_size=(3, 3), activation="relu"),
        layers.MaxPooling2D(pool_size=(2, 2)),
        layers.Flatten(),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation="softmax"),
    ]
)
model.summary()

batch_size = 128
epochs = 15
model.compile(loss="categorical_crossentropy", optimizer   ="adam", metrics=["accuracy"])
model.fit(x_train, y_train, batch_size=batch_size, epochs   =epochs, validation_split=0.1)

score = model.evaluate(x_test, y_test, verbose=0)
print("Test loss:", score[0])
print("Test accuracy:", score[1])

```

LeNet-5

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np

# 1) MNIST 데이터 로드
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# LeNet 입력 형태에 맞게 reshape (28×28 → 28×28×1)
x_train = x_train.reshape(-1, 28, 28, 1).astype('float32') / 255.0
x_test  = x_test.reshape(-1, 28, 28, 1).astype('float32') / 255.0

# 2) LeNet-5 모델 정의
model = models.Sequential([
    layers.Conv2D(6, (5, 5), activation='tanh', input_shape=(28, 28, 1)),
    layers.AveragePooling2D(2,2),
    layers.Conv2D(16, (5, 5), activation='tanh'),
    layers.AveragePooling2D(2,2),
    layers.Flatten(),
    layers.Dense(120, activation='tanh'),
    layers.Dense(84, activation='tanh'),
    layers.Dense(10, activation='softmax')
])

# 3) 모델 컴파일 및 학습
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()

history = model.fit(
    x_train, y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.1
)

# 4) 테스트셋 정확도 평가
test_loss, test_acc = model.evaluate(x_test, y_test)
print("Test accuracy:", test_acc)

# 5) 예측 테스트
sample = x_test[0].reshape(1, 28, 28, 1)
pred = model.predict(sample)
print("Predicted:", np.argmax(pred))
plt.imshow(x_test[0], cmap="Greys")
```

EfficientNetV2

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import EfficientNetV2S
import matplotlib.pyplot as plt
import numpy as np

# 1데이터셋 로드 및 전처리 (CIFAR-10)
print("데이터를 로드 중입니다...")
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# EfficientNetV2S 권장 입력 크기에 맞게 조정 (메모리 사양에 따라 128~224 권장)
IMG_SIZE = 128
CHANNELS = 3
NUM_CLASSES = 10

def preprocess_data(images, labels):
    # 이미지 리사이징 및 배치 처리용 함수
    dataset = tf.data.Dataset.from_tensor_slices((images, labels))
    dataset = dataset.map(lambda x, y: (tf.image.resize(x, (IMG_SIZE, IMG_SIZE)), y))
    dataset = dataset.shuffle(1000).batch(32).prefetch(tf.data.AUTOTUNE)
    return dataset

train_ds = preprocess_data(x_train, y_train)
test_ds = preprocess_data(x_test, y_test)

# 2. EfficientNetV2S 모델 빌드
def build_efficientnet_v2(num_classes):
    # Imagenet 가중치를 사용하고 최상단 레이어(분류기)는 제외
    base_model = EfficientNetV2S(
        weights='imagenet',
        include_top=False,
        input_shape=(IMG_SIZE, IMG_SIZE, CHANNELS)
    )

    # 특징 추출기(Base Model)의 가중치는 고정 (학습 속도 향상)
    base_model.trainable = False

    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.BatchNormalization(),
        layers.Dropout(0.3),  # 과적합 방지
        layers.Dense(num_classes, activation='softmax')
    ])

    return model

model = build_efficientnet_v2(NUM_CLASSES)

model.compile( optimizer='adam',loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()

print("\n학습을 시작합니다. (코랩 GPU 모드를 권장합니다)")
epochs = 5 # 실습용으로 5회 설정
history = model.fit(train_ds, epochs=epochs,validation_data=test_ds)

# 5. 결과 시각화
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(epochs)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, val_acc, label='Validation Accuracy')
plt.title('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, val_loss, label='Validation Loss')
plt.title('Loss')
plt.legend()
plt.show()
```

```python
class_names = ['비행기', '자동차', '새', '고양이', '사슴', '개', '개구리', '말', '배', '트럭']

# CIFAR-10 테스트셋에서 6번째 이미지를 추출 (원본 데이터 사용)
test_image_original = x_test[6]  # 원본 32x32 이미지
test_true_label_idx = y_test[6][0]
test_true_label = class_names[test_true_label_idx]

# 테스트 이미지를 모델 입력 형태로 전처리 (동일한 전처리 파이프라인 사용)
def preprocess_single_cifar(image):
    """단일 CIFAR-10 이미지 전처리"""
    image = tf.cast(image, tf.float32) / 255.0
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = tf.expand_dims(
        tf.keras.applications.mobilenet_v3.preprocess_input(image * 255.0), 0
    )
    return image

# 0번째 테스트 이미지 전처리
test_image_processed = preprocess_single_cifar(test_image_original)

# 예측 수행
predictions = model.predict(test_image_processed, verbose=0)
predicted_class = np.argmax(predictions[0])
confidence = np.max(predictions[0])
predicted_label = class_names[predicted_class]

print(f"   예측 클래스: {predicted_label}")
print(f"   신뢰도: {confidence:.1%}")
print(f"   실제 클래스: {test_true_label}")


# 테스트 결과 시각화
plt.figure(figsize=(15, 5))

# 1) 원본 CIFAR-10 이미지 (32x32)
plt.subplot(1, 3, 1)
plt.imshow(test_image_original)
plt.title(f'원본 CIFAR-10 (32x32)\n실제: {test_true_label}', fontsize=14)
plt.axis('off')
```

MobileNetV3

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV3Small  # EfficientNetV2S → MobileNetV3Small
import matplotlib.pyplot as plt
import numpy as np

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# MobileNetV3 권장 입력 크기에 맞게 조정 (128x128 권장)
IMG_SIZE = 128
CHANNELS = 3
NUM_CLASSES = 10

def preprocess_data(images, labels):
    # MobileNetV3 전처리 추가 (EfficientNet과 다름)
    dataset = tf.data.Dataset.from_tensor_slices((images, labels))
    dataset = dataset.map(
        lambda x, y: (
            tf.keras.applications.mobilenet_v3.preprocess_input(
                tf.image.resize(tf.cast(x, tf.float32)/255.0, (IMG_SIZE, IMG_SIZE)) * 255.0
            ),
            y
        ),
        num_parallel_calls=tf.data.AUTOTUNE
    )
    dataset = dataset.shuffle(1000).batch(32).prefetch(tf.data.AUTOTUNE)
    return dataset

train_ds = preprocess_data(x_train, y_train)
test_ds = preprocess_data(x_test, y_test)

# MobileNetV3Small 모델 빌드
def build_mobilenetv3(num_classes):
    # Imagenet 가중치를 사용하고 최상단 레이어(분류기)는 제외
    base_model = MobileNetV3Small(  # 변경된 부분
        weights='imagenet',
        include_top=False,
        input_shape=(IMG_SIZE, IMG_SIZE, CHANNELS)
    )

    # 특징 추출기(Base Model)의 가중치는 고정 (학습 속도 향상)
    base_model.trainable = False

    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.BatchNormalization(),
        layers.Dropout(0.3),  # 과적합 방지
        layers.Dense(num_classes, activation='softmax')
    ])

    return model

model = build_mobilenetv3(NUM_CLASSES)

model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
model.summary()

print("\n학습을 시작합니다. (코랩 GPU 모드를 권장합니다)")
epochs = 5 #  5회 설정
history = model.fit(train_ds, epochs=epochs, validation_data=test_ds)

# 결과 시각화
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(epochs)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, val_acc, label='Validation Accuracy')
plt.title('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, val_loss, label='Validation Loss')
plt.title('Loss')
plt.legend()
plt.show()
```

```python
class_names = ['비행기', '자동차', '새', '고양이', '사슴', '개', '개구리', '말', '배', '트럭']

# CIFAR-10 테스트셋에서 6번째 이미지를 추출 (원본 데이터 사용)
test_image_original = x_test[6]  # 원본 32x32 이미지
test_true_label_idx = y_test[6][0]
test_true_label = class_names[test_true_label_idx]

print(f"실제 레이블: {test_true_label} (인덱스: {test_true_label_idx})")

# 테스트 이미지를 모델 입력 형태로 전처리 (동일한 전처리 파이프라인 사용)
def preprocess_single_cifar(image):
    """단일 CIFAR-10 이미지 전처리"""
    image = tf.cast(image, tf.float32) / 255.0
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = tf.expand_dims(
        tf.keras.applications.mobilenet_v3.preprocess_input(image * 255.0), 0
    )
    return image

# 0번째 테스트 이미지 전처리
test_image_processed = preprocess_single_cifar(test_image_original)

# 예측 수행
predictions = model.predict(test_image_processed, verbose=0)
predicted_class = np.argmax(predictions[0])
confidence = np.max(predictions[0])
predicted_label = class_names[predicted_class]

print(f"\n MobileNetV3 예측 결과:")
print(f"   예측 클래스: {predicted_label}")
print(f"   신뢰도: {confidence:.1%}")
print(f"   실제 클래스: {test_true_label}")


# 테스트 결과 시각화
plt.figure(figsize=(15, 5))

# 1) 원본 CIFAR-10 이미지 (32x32)
plt.subplot(1, 3, 1)
plt.imshow(test_image_original)
plt.title(f'원본 CIFAR-10 (32x32)\n실제: {test_true_label}', fontsize=14)
plt.axis('off')
```

```python
import numpy as np
from tqdm import tqdm

# --- GPT-2 연산 함수들 ---
def gelu(x):
    return 0.5 * x * (1 + np.tanh(np.sqrt(2 / np.pi) * (x + 0.044715 * x**3)))

def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)

def layer_norm(x, g, b, eps: float = 1e-5):
    mean = np.mean(x, axis=-1, keepdims=True)
    variance = np.var(x, axis=-1, keepdims=True)
    return g * (x - mean) / np.sqrt(variance + eps) + b

def linear(x, w, b):
    return x @ w + b

def ffn(x, c_fc, c_proj):
    return linear(gelu(linear(x, **c_fc)), **c_proj)

def attention(q, k, v, mask):
    return softmax(q @ k.T / np.sqrt(q.shape[-1]) + mask) @ v

def mha(x, c_attn, c_proj, n_head):
    x = linear(x, **c_attn)
    qkv_heads = list(map(lambda x: np.split(x, n_head, axis=-1), np.split(x, 3, axis=-1)))
    causal_mask = (1 - np.tri(x.shape[0], dtype=x.dtype)) * -1e10
    out_heads = [attention(q, k, v, causal_mask) for q, k, v in zip(*qkv_heads)]
    x = linear(np.hstack(out_heads), **c_proj)
    return x

def transformer_block(x, mlp, attn, ln_1, ln_2, n_head):
    x = x + mha(layer_norm(x, **ln_1), **attn, n_head=n_head)
    x = x + ffn(layer_norm(x, **ln_2), **mlp)
    return x

def gpt2(inputs, wte, wpe, blocks, ln_f, n_head):
    x = wte[inputs] + wpe[range(len(inputs))]
    for block in blocks:
        x = transformer_block(x, **block, n_head=n_head)
    return layer_norm(x, **ln_f) @ wte.T

def generate(inputs, params, n_head, n_tokens_to_generate):
    for _ in tqdm(range(n_tokens_to_generate), "Generating"):
        logits = gpt2(inputs, **params, n_head=n_head)
        next_id = np.argmax(logits[-1])
        inputs.append(int(next_id))
    return inputs[len(inputs) - n_tokens_to_generate :]

# --- 실행 함수 ---
def run_gpt2(prompt, n_tokens_to_generate=40, model_size="124M", models_dir="models"):
    from utils import load_encoder_hparams_and_params

    # 모델 가중치 로드 (picoGPT 구조 기준)
    encoder, hparams, params = load_encoder_hparams_and_params(model_size, models_dir)

    input_ids = encoder.encode(prompt)
    output_ids = generate(input_ids, params, hparams["n_head"], n_tokens_to_generate)
    output_text = encoder.decode(output_ids)

    print(f"\n[입력 문장]: {prompt}")
    print(f"[생성 결과]: {output_text}")
    return output_text

# 코랩에서 즉시 실행
results = run_gpt2(prompt="Alan Turing was a", n_tokens_to_generate=20)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.init as init
import torchvision.datasets as dset
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
import numpy as np
import matplotlib.pyplot as plt
print(torch.__version__)

batch_size = 256
learning_rate = 0.0002
num_epoch = 10

mnist_train = dset.MNIST(root="../", train=True, transform=transforms.ToTensor(), target_transform=None,
download=True)
mnist_test = dset.MNIST(root="../", train=False, transform=transforms.ToTensor(), target_transform=None,
download=True)

train_loader = DataLoader(mnist_train,batch_size=batch_size, shuffle=True,num_workers=2,drop_last=True)
test_loader = DataLoader(mnist_test,batch_size=batch_size, shuffle=False,num_workers=2,drop_last=True)    #data loader 설정

class CNN(nn.Module):
    def __init__(self):
        super(CNN,self).__init__()
        self.layer = nn.Sequential(
            nn.Conv2d(in_channels=1,out_channels=16,kernel_size=5),          # [batch_size,1,28,28] -> [batch_size,16,24,24]
            nn.ReLU(),                                                                                  # 필터의 개수는 1개에서 16개로 늘어남
            nn.Conv2d(in_channels=16,out_channels=32,kernel_size=5),        # [batch_size,16,24,24] -> [batch_size,32,20,20]
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2,stride=2),                                         # [batch_size,32,20,20] -> [batch_size,32,10,10]
            nn.Conv2d(in_channels=32, out_channels=64, kernel_size=5),       # [batch_size,32,10,10] -> [batch_size,64,6,6]
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2,stride=2)                                           # [batch_size,64,6,6] -> [batch_size,64,3,3]
        )
        self.fc_layer = nn.Sequential(
            nn.Linear(64*3*3,100),                                                # [batch_size,64*3*3] -> [batch_size,100]
            nn.ReLU(),
            nn.Linear(100,10)                                                         # [batch_size,100] -> [batch_size,10]
        )

    def forward(self,x):
        out = self.layer(x)
        out = out.view(x.size(0),-1)                  # view 함수가 [batch_size,나머지]로 바꿔줌
        out = self.fc_layer(out)
        return out

from torchsummary import summary

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu") # gpu가 사용 가능한 경우에는 device를 gpu로 설정
print(device)
model = CNN().to(device) # 모델을 장치로 올림
loss_func = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
print(model) #ahs


summary(model,(1,28,28)) #모델과 파라미터 보기, 대부분 Trainable params임

loss_arr =[]
for i in range(num_epoch):    	# 학습시작
    for j,[image,label] in enumerate(train_loader):
        x = image.to(device)                	# 입력 데이터를 GPU 로 이동
        y_= label.to(device)
        optimizer.zero_grad()  		#기울기를 초기화
        output = model.forward(x)  	#출력계산
        loss = loss_func(output,y_)  	#손실함수 계산
        loss.backward()                     	#역전파 수행
        optimizer.step()                      	# 옵티마이저를 사용하여 모델의 파라미터를 업데이트

        if j % 1000 == 0:
            print(loss)
            loss_arr.append(loss.cpu().detach().numpy())
             			# 현재 손실 값을 loss_arr 배열에 추가

plt.plot(loss_arr)
plt.show()
```

```python
correct = 0
total = 0

with torch.no_grad():                 	#  테스트 단계에서 자동 미분을 비활성화
    for image,label in test_loader:
        x = image.to(device)         	# 입력 데이터를 GPU로 이동
        y_= label.to(device)
        output = model.forward(x) 	# 모델에 입력 데이터를 넣어 예측을 생성

        _,output_index = torch.max(output,1)   	# 모델의 출력에서 가장 높은 값을 인덱스화
        total += label.size(0)
        correct += (output_index == y_).sum().float()  	# 정확하게 예측한 데이터의 개수를 누적

    print('Accuracy of Test Data:', (100*correct/total))
```

```python
import tensorflow as tf
from tensorflow.keras import models, layers
import matplotlib.pyplot as plt

# MNIST 데이터 불러오기
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# 0~1로 정규화
train_images = x_train / 255.0
test_images = x_test / 255.0

# 정수 레이블을 One-Hot Encoding으로 변환
y_train = tf.keras.utils.to_categorical(y_train, 10)
y_test = tf.keras.utils.to_categorical(y_test, 10)

# 모델 생성
model = models.Sequential()

model.add(layers.Flatten(input_shape=(28, 28)))   # 28x28 → 784차원
model.add(layers.Dense(10, activation='sigmoid')) # 출력층 10개

# SGD + MSE
model.compile(   optimizer='sgd',  loss='mean_squared_error',  metrics=['accuracy'])

# 학습
history = model.fit(  train_images,  y_train, epochs=5)

# 테스트
test_loss, test_acc = model.evaluate(  test_images, y_test)

print('테스트 정확도:', test_acc)

# 학습 결과
loss = history.history['loss']
acc = history.history['accuracy']

epochs = range(1, len(loss) + 1)

plt.plot(epochs, loss, 'b', label='Training Loss')
plt.plot(epochs, acc, 'r', label='Accuracy')

plt.xlabel('epochs')
plt.ylabel('loss / accuracy')
plt.legend()
plt.show()

# 첫 번째 테스트 이미지 예측
pred = model.predict( test_images[0].reshape(1, 28, 28), batch_size=1)

print("추정된 숫자 =", pred.argmax())

# 첫 번째 이미지 출력
plt.imshow(test_images[0], cmap="Greys")
plt.show()
```

```python

import tensorflow as tf
from tensorflow.keras import models, layers
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
train_images = x_train / 255.0
test_images = x_test / 255.0

model = models.Sequential()
model.add(layers.Flatten(input_shape=(28, 28)))  # 784차원 변환
model.add(layers.Dense(10, activation='sigmoid'))  # 출력층(단층)

model.compile(optimizer='adam',loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history=model.fit(train_images, y_train, epochs=5)

test_loss, test_acc = model.evaluate(test_images,  y_test)
print('테스트 정확도:', test_acc)

loss = history.history['loss']
acc = history.history['accuracy']
epochs = range(1, len(loss)+1)

plt.plot(epochs, loss, 'b', label='Training Loss')
plt.plot(epochs, acc, 'r', label='Accuracy')
plt.xlabel('epochs')
plt.ylabel('loss/acc')
plt.show()
#0~9999
pred = model.predict(test_images[0].reshape(1,28,28,1), batch_size=1)
print("추정된 숫자=", pred.argmax())
plt.imshow(test_images[0], cmap="Greys")
```

```python
import tensorflow as tf
from tensorflow.keras import models, layers
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_train = x_train / 255.0
x_test = x_test / 255.0

model = models.Sequential()
model.add(layers.Flatten(input_shape=(28, 28)))  # 784차원 변환
model.add(layers.Dense(128, activation='relu'))   # 은닉층 1
model.add(layers.Dense(64, activation='relu'))    # 은닉층 2
model.add(layers.Dense(10, activation='softmax'))  # 출력층(단층)

model.compile(optimizer='adam',loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history=model.fit(x_train, y_train, epochs=5)

test_loss, test_acc = model.evaluate(x_test, y_test)
print('테스트 정확도:', test_acc)

loss = history.history['loss']
acc = history.history['accuracy']
epochs = range(1, len(loss)+1)

plt.plot(epochs, loss, 'b', label='Training Loss')
plt.plot(epochs, acc, 'r', label='Accuracy')
plt.xlabel('epochs')
plt.ylabel('loss/acc')
plt.show()

pred = model.predict(test_images[0].reshape(1,28,28,1), batch_size=1)
print("추정된 숫자=", pred.argmax())
plt.imshow(test_images[0], cmap="Greys")
```
